### Decision Tree

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import classification_report
```

```python

df=pd.read_csv('loan_approval_dataset.csv')
df
```

```python

x=df.drop(columns='Loan_Approved')
y=df.Loan_Approved
```

```python
obj_cols=x.select_dtypes(include='object').columns
obj_cols
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
xtrain[obj_cols].nunique()
```

```python

encode=OneHotEncoder(sparse_output=False, handle_unknown='ignore')
values=encode.fit_transform(xtrain[obj_cols])
cols=encode.get_feature_names_out()
```

```python
xtrain[cols]=values
```

```python

xtrain.drop(columns=obj_cols,inplace=True)
xtrain
```

```python
v=encode.transform(xtest[obj_cols])
c=encode.get_feature_names_out()
xtest[c]=v
xtest.drop(columns=obj_cols,inplace=True)
xtest
```

```python
model=DecisionTreeClassifier()
model.fit(xtrain,ytrain)
```

```python

ypred_train=model.predict(xtrain)
```

```python
print(classification_report(ytrain,ypred_train))
```

```python
ypred_test=model.predict(xtest)
```

```python

print(classification_report(ytest,ypred_test))
```

### Conclusion
- For train data the accuracy is 100%, hence the learned the patterns well on seen data
- But for test data ,the accuracy is 52% which shows the model performs better only on seen data, not on unseen data. Hence model is not generalized well on unseen data
